```python
## 
import PyPDF2
import pyttsx3
from tkinter.filedialog import *

def readPdfFile():
    paper= askopenfilename()
    fileread = PyPDF2.PdfFileReader(paper)
    pagecounts = fileread.numPages

    for i in range(1,pagecounts-1):
        page = fileread.getPage(i)
        text = page.extractText()
        audio = pyttsx3.init()
        """ RATE"""
        audio.setProperty('rate', 500)     # setting up new voice rate
        """VOICE"""
        voices = audio.getProperty('voices')       #getting details of current voice
        audio.setProperty('voice', voices[1].id) 
        audio.say(text)
    audio.runAndWait()
        
readPdfFile()
```
