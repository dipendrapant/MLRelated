```python
from bs4 import BeautifulSoup
import requests
n = int(input("Enter no of pages = "))
output_urls = []
for i in range(1,n+1):
  url = BeautifulSoup(f'https://english.onlinekhabar.com/category/political/page/{i}', 'html.parser')
  response = requests.get(url)
  data = response.text
  soup = BeautifulSoup(data,'lxml')
  soup = soup.find_all("div",'ok-news-post ltr-post')
  n = len(soup)
  news_urls = []
  for i in range(n):
    news_url = soup[i].find('a').get('href')
    news_urls.append(news_url)
  
  output_urls.append(news_urls)
```

```python
news_url_flat = []
for i in news_urls:
#     for j in i:
    news_url_flat.append(i)
```

```python
from bs4 import BeautifulSoup

url = BeautifulSoup('https://english.onlinekhabar.com/no-amnesty-for-those-involved-in-serious-human-rights-violation-gyawali.html', 'html.parser')
res = requests.get(url)

html_page = res.content
soup = BeautifulSoup(html_page, 'lxml')
text = soup.find("div",'post-content-wrap')
text = text.find_all(text=True)

output = ''
for t in text:
    output += '{} '.format(t)

print(output)
```
