```python
from bs4 import BeautifulSoup
import requests
```

```python
url = BeautifulSoup('https://english.onlinekhabar.com/category/political','html.parser')
response = requests.get(url)
print(response)
```

```python
data = response.text
data
```

```python
soup = BeautifulSoup(data,'lxml')
soup
```

```python
soup = soup.find("div","ok-news-post ltr-post")
soup
```

```python
soup.find('a').get('href')
```

```python
n = len(soup)
news_urls = []
for i in range(n):
    news_url = soup[i].find('a').get('href')
    news_urls.append(news_url)
```
